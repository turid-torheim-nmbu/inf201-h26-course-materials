## Dice class

The `Dice` class allows us to introduce randomness. A dice is created with a given number of sides (min=2, default=6), and has methods for rolling and calling the dice.

```python
d20 = Dice(sides=20)
d20
```

```python
d20.roll()
```

```python
d20()  # implemented using the __call__ special method
```

## Weapons class

Weapons are in the `items` module. A weapon has a damage dice that is used to roll for damage. A `Character` object can hold a weapon, and use it to attack. A `Character` object can equip a weapon using the `Character.equip_weapon()` method.

```python
d12 = Dice(12)
sword2 = Weapon("Anduril", d12)
sword2
```

```python
sword2.roll_damage()
```

## Simulating battles

The `Battle` class allows two characters to fight intil one of them wins. We will no simulate battles between different combinations of characters.

Note: In this example solution we have implemented a more detailed log in the `Battle` and `Character` classes to get a more interesting output.

### Warrior vs Mage - no weapons

```python
warrior = Warrior("Thorin", hp_max=80, damage=10, armor=3)
mage = Mage("Gandalf", hp_max=60, damage=12, mana_max=20)

battle = Battle(warrior, mage)
winner = battle.simulate_battle(verbose=True)
```

### Character vs Character - no weapons

```python
hero = Character("Merry", hp_max=75, damage=15)
enemy = Character("Orc", hp_max=50, damage=20)

battle = Battle(hero, enemy)
winner = battle.simulate_battle(verbose=True)
```

### Character vs Character - with weapons

```python
d6 = Dice(sides=6)
w1 = Weapon("Sword", d6)

d10 = Dice(sides=10)
w2 = Weapon("Axe", d10)

hero = Character("Merry", hp_max=75, damage=15, weapon=w1)
enemy = Character("Orc", hp_max=50, damage=20, weapon=w2)

battle = Battle(hero, enemy)
winner = battle.simulate_battle(verbose=True)
```

### Warrior vs Mage - with weapons

```python
d12 = Dice(12)
w3 = Weapon("Staff", d12)

warrior = Warrior("Thorin", hp_max=80, damage=10, armor=3, weapon=w2)
mage = Mage("Gandalf", hp_max=60, damage=12, mana_max=20, weapon=w3)

battle = Battle(warrior, mage)
winner = battle.simulate_battle(verbose=True)
```
