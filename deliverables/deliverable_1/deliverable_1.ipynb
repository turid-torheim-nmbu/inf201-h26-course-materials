### Task 1: Temperature
* First request the daily mean air temperture for all of 2025.
    * The specifications you will need for this is:
        ```
        endpoint = 'https://frost.met.no/observations/v0.jsonld'
        parameters = {
            'sources': 'SN17850',
            'elements': 'mean(air_temperature P1D)',
            'referencetime': '2025-01-01/2025-12-31',
        }
    * Otherwise you can follow the example from frost.met.no on how to get the data when using Python
* Make a nicely formatted plot of this air temperature, where time is on the x-axis and temperature is on the y-axis
* Create a nicely formatted summary table containing mean, median, min and max temperature for this year, and display it with print()

### Task 2: Precipitation
* Request data for precipitation for the same time period, specifically measurements of precipitation over 24 hours. This is coded as `sum(precipitation_amount P1D)` in the database
* Create a new plot that shows both the temperature from task 1 and the precipitation in the same plot
    * Think through how to display both values in a way that is useful and looks good
    * Consider which type of plot works best for each measurement (scatter, line, bar, pie chart, ...?)

### Task 3: Historical temperature data
* Now you want to compare the 2025 temperatures to temperatures measured 100 years earlier.
* Get these temperatures from the database.
* Plot the new and the old temperatures in the same plot, still nicely formatted
* Generate a summary table as for task 1, where you display data from both the new and the old temperatures.

### Task 4: Heatwaves
* The next task is to look for heatwaves in Ås in 2025
* In Norway a heatwave is defined as a consecutive period of five or more day where the<br>
<i>max</i> temperature is >= 27 degrees Celcius.
    * Note that you need max temperature here, not mean
    * Max temperature is coded as `max(air_temperature P1D)`
    * [met.no article about heatwaves](https://www.met.no/vaer-og-klima/ekstremvaervarsler-og-andre-farevarsler/vaerfenomener-som-kan-gi-farevarsel-fra-met/hetebolger) (only in Norwegian)
* Write a function that count the number of heatwaves in a year.
    * The function should write a summary to a yml file, formatted as the example file
    * Write the function so that it will work with any station name/id and year

### Task 5: Moving average temperature
* Write a function that calculates the 7-day moving average for temperature
* Use this function on the 2025 mean temperature you collected in task 1
* Make a plot showing the moving average as a line and the daily temperature as points
