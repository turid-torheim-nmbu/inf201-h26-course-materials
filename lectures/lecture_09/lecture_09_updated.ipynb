## Composition and inheritance
* These both indicate how two objects relate to each other.
* Compsition: Object a *has* an object b.
    * Composition represents a "has-a" relationship between objects. (from [here](https://www.geeksforgeeks.org/python/inheritance-and-composition-in-python/))
* Inheritance: Class C1 *is* a (subclass of) class C .
    * Inheritance represents an "is-a" relationship between classes. (from [here](https://www.geeksforgeeks.org/python/inheritance-and-composition-in-python/))
* Let's create a couple of classes to demontrate these two concepts.

````python
class Engine:
    """Represents an engine."""

    def __init__(self, horsepower: int = 150) -> None:
        """Create engine with given power"""
        self.horsepower = horsepower

    def start(self) -> str:
        """Create string indicating starting the engine"""
        return f"Engine rumbles to life with {self.horsepower} HP!"
````

````python
class Vehicle:
    """Base vehicle that delegates engine behavior."""

    def __init__(self, brand: str, engine: Engine) -> None:
        """Create a vehicle of specified brand with engine"""
        self.brand = brand
        self.engine = engine  # COMPOSITION: Vehicle has an Engine

    def start_journey(self) -> list[str]:
        """Create list of strings indicating the vehicle has started"""
        engine_msg = self.engine.start()  # Delegate action to the engine component
        return [engine_msg, f"{self.brand} is on the road."]
````

````python
e1 = Engine(horsepower=200)
v1 = Vehicle("Ford", e1)
v1.start_journey()
````

* Here we have an example of **composition**: The Vehicle has an Engine.
    * In `Vehicle.start_journey()` the `Vehicle` class delegates `self.engine.start()` to the `Engine` class
    * If we implement changes to the `Engine` class (like including a fuel level check) does not require us to make any changes in `Vehicle`

````python
class Truck(Vehicle):
    """Specialized vehicle using inheritance."""

    def __init__(
        self, brand: str, engine: Engine, max_payload_tons: float = 5.0
    ) -> None:
        # Pass base attributes up to parent constructor
        super().__init__(brand=brand, engine=engine)
        self.max_payload_tons = max_payload_tons

    def start_journey(self) -> list[str]:
        # Extend parent behavior
        logs = [f"Safety check passed for {self.max_payload_tons}t payload."]
        logs.extend(super().start_journey())
        return logs
````

* This is **inheritance**: `Truck` is a subclass of `Vehicle`
    * `Truck` re-uses the attributes `brand` and `engine` from Vehicle
    * Any attribute or method implemented for `Vehicle` will also be available to `Truck`
    * The sub-class can overwrite or extend class methods if needed, as done here for `start_journey()`

| Feature | Inheritance | Composition |
|:-|:-|:-|
| Relationship | Is-A | Has-A |
| Reuse mechanism | Inherits class behavior | Uses another object |
| Dependency | Child depends on parent hierarchy | Container depends on contained object |
| Behavior change | Overriding or extending inherited methods | Delegation or replacing contained objects |
| Class hierarchy | Creates hierarchy | Avoids hierarchy |
| Flexibility | Less flexible for changing relationships | More flexible through interchangeable objects |
| Best suited for | Closely related class types | Combining independent responsibilities |

````python
class SubClass(ClassName):
    """Subclass of ClassName"""

    def __init__(self, required_param: str, default_param: int = 10, sub_class_param: bool = False) -> None:
        """Initialise sub-class object"""
        super().__init__(required_param=required_param, default_param=default_param)
        self.sub_class_attribute = sub_class_param

    def add_numbers(self):
        """Add 10 to default parameter"""
        return self.attribute2 + 10

    def class_text(self) -> str:
        """Returns string with class info"""
        txt_sc = "Subclass: SubClass. "
        txt_from_class = super().class_text()
        return txt_sc + txt_from_class
````

## Public, protected and private attributes
* We talked a bit about public and protected attributes last week
* Public attributes
    * Syntax: No leading underscores (`self.name`).
    * Behavior: Accessible and modifiable from anywhere—inside the class, by subclasses, and from outside instance calls.
    * When to use: For attributes and methods that are intended to be part of the class's public interface.
* Protected attributes
    * Syntax: A single leading underscore (`self._name`).
    * Behavior: Serves as a convention telling other developers, "This is internal logic/data—do not touch it directly outside the class unless you are writing a subclass." Python does *not* strictly enforce this at runtime. The code will still run if accessed externally, but tools like VS Code/Pylance will flag warnings.
* Private attributes
    * Syntax: Double leading underscores without trailing underscores (`self.__name`).
    * Behavior: Triggers name mangling. Python automatically rewrites the attribute name under the hood by prepending `_ClassName` (such as `_ClassName__name`). This prevents accidental overwriting or external access using the original name.
    * When to use: To avoid name collisions in subclassing, or to make accidental direct access noticeably harder.
    

| Access Modifier | Python Syntax | Access from Outside Class | Access from Subclass | Enforcement Level |
|:-|:-|:-|:-|:-|
| Public | `self.name` | Allowed | Allowed | None (Open access) |
| Protected | `self._name` | Discouraged | Allowed | Convention only (Linter warnings) |
| Private | `self.__name` | Blocked via standard name | Blocked via standard name | Name mangling (`_ClassName__name`) |

* Why do we not use private attributes more?
    * It breaks inheritance because the attribute name is strictly bound to the class where it was defined
        ```python
        class Character:
            """A game character"""

            def __init__(self, hp: int) -> None:
                """Initialise character with HP"""
                self.__hp = hp  # Mangled to _Character__hp


        class Warrior(Character):
            """A warrior character"""

            def heal(self, amount: int) -> None:
                """Heal for amount HP"""
                self.__hp += amount  # Fails with AttributeError: 'Warrior' object has no attribute '_Warrior__hp'
        ```
    * It provides a false sense of security
        ```python
            ch1 = Character(hp=25)
            print(ch1.__hp)  # Fails with AttributeError: 'Character' object has no attribute '__hp'
            print(ch1._Character__hp)  # Can still be accessed using the mangled name
            ch1._Character__hp = -999  # Can still be changed using the mangled name
        ```
    * Code is harder to read, test and debug
    * Python's community standards expect you to respect the protected attribute (`_name`), making the need for a private attribute (`__name`)
    * Use a single underscore (_) by default whenever an attribute or method is internal. Only use double underscores (__) if you are building deep inheritance structures where name collisions with unknown subclasses become a real risk.


````python
class Character:
    """A game character"""

    def __init__(self, hp: int) -> None:
        """Initialise character with HP"""
        self.__hp = hp  # Mangled to _Character__hp


ch1 = Character(hp=25)
print(f"Using mangled name: {ch1._Character__hp}")
print(f"Using regular name: {ch1.__hp}")
````

* `__getitem__`
    * Allows index and key look-up, such as `class_object[0]` or `class_object["key_word"]`
    * Allows slicing, such as `class_object[0:2]`
* `__contains__`
    * Does the object contain the specified item? `item in class_object`

````python
# See file dataset_with_type_hints.py for version with full type hints and doc-strings
class Dataset:

    def __init__(self, records: list) -> None:
        self._records = records

    def __len__(self) -> int:
        return len(self._records)

    def __getitem__(self, index):
        # Enables d[0], d[1:3], and even iteration for free!
        return self._records[index]

    def __contains__(self, item) -> bool:
        # Enables 'item in dataset'
        return item in self._records
````

### Methods for comparisons
* `__eq__`
    * Customises the `==` operator
    * What does it mean for two objects to be considered equal?
    * If not implemented, `==` will default to checking object identity (memory address)
    * `True`, `False` or `NotImplemented`
* `__lt__`
    * Customises the `<` operator
    * How do we check if this object is less than another object?
* `__le__`, `__gt__`, and `__ge__`
    * Customises `<=`, `>` and `>=`
    * All about comparison of objects
* By using `total_ordering` from `functools`, you can define `__eq__` and one of the other comparisons in the class, and the rest will be implemented by total_ordering
    * `@total_ordering` is another example of a decorator, like `@property` we used last week

````python
from functools import total_ordering


@total_ordering
class Vector:

    def __init__(self, x: float, y: float):
        self.x = x
        self.y = y

    def __eq__(self, other) -> bool:
        if not isinstance(other, Vector):
            return NotImplemented
        return self.x == other.x and self.y == other.y

    def __lt__(self, other) -> bool:
        if not isinstance(other, Vector):
            return NotImplemented
        return (self.x**2 + self.y**2) < (other.x**2 + other.y**2)
````

### Methods for context management
* `__enter__` and `__exit__`
    * For working with contexts (`with`)
    * Determines what happens when we enter and exit the context
    * More about creating context managers [here](https://www.pythonmorsels.com/creating-a-context-manager/)

````python
import time


class Timer:

    def __enter__(self):
        self.start = time.perf_counter()
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        self.elapsed = time.perf_counter() - self.start
        print(f"Execution took: {self.elapsed:.4f} seconds")
        return False  # Do not suppress exceptions
````

````python
with Timer():
    # Code block being timed
    sum(i**2 for i in range(10_000_000))
````

### Method for making an object callable
* Allows you to call the object as if it was a function

````python
class Counter:

    def __init__(self):
        self.count = 0

    def __call__(self):
        self.count += 1
        return self.count


# Create a counter object
my_counter = Counter()
# Call the object multiple times
print(my_counter())  # Output: 1
print(my_counter())  # Output: 2
print(my_counter())  # Output: 3
````

### Why do objects have methods that we have not implemented?
* Every class in Python implicitly inherits from the `object` class
    ```python
        class NewClass():  # What we write
            pass

        class NewClass(object):  # What happens behind the scene
            pass
    ```
* We used this last week when we looked at how everything is an object
    ```python
        print(isinstance(42, object))  # True
        print(isinstance("hello", object))  # True
        print(isinstance([1, 2, 3], object))  # True
        print(isinstance(NewClass, object))  # True (even classes are objects!)
    ```
