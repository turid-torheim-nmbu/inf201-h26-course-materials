## Formatted string literals: f-strings
* A very flexible way to format strings in Python

````python
# Printing or writing nice looking strings with a lot of flexibility
name = "Turid"
course = "INF201"
year = 2026
print("Hello, " + name + "! You are teaching " + course + " the autumn of " + str(year) + ".")
print(f"Hello, {name}! You are teaching {course} the autumn of {year}.")

# Formatting numbers in strings
number = 2
print(f"{number} divided by 3 is {number / 3}.")
print(f"{number} divided by 3 is {number / 3:.2f}.")
print(f"{number} divided by 3 is {number / 3:.1%}.")

# Alignment and padding
print(f"{42:0>5}")
print(f"{'Lower case':<15}{'Upper case':<15}")
for i in range(5):
    lower = chr(97 + i)
    upper = chr(65 + i)
    print(f"{lower:<15}{upper:<15}")
````

## Dynamic vs static typing

### Python is dynamically typed
* We do not have to tell Python which data type a variable has
* Type is inferred when we run the code
* A variable can change type without problems
    ```python
    a = 3  # a in an integer
    a = "text"  # a is now a string
    ```
* This would not be the case in a statically typed language
    ```c
    // C code
    int x = 4;
    ```

### Comparison of static and dynamic typing
* Dynamic typing
    * Examples: Python, Ruby
    * Type is determined at run-time
    * A variable is a label or reference to an object in memory 
    * Faster and easier to write
    * Unexpected type errors when running code
* Static typing
    * Examples: C, Java
    * Type is determined at compile-time
    * Variables are typed “boxes” (locations) in memory
    * Type errors are caught during compilation
    * Code runs faster
* The Whilwind Tour of Python has a [good chapter](https://jakevdp.github.io/WhirlwindTourOfPython/03-semantics-variables.html) introducing this concept
* (https://nedbatchelder.com/text/names) is a excellent and more thorough discussion of this (read it!)

## Beware of the aliasing trap   
    

````python
x = [1, 2, 3]
y = x
y.append(4)
print(y)  # Output: [1, 2, 3, 4]
print(x)
print(id(x))
print(id(y))
print(id(x) == id(y))
````

* `x` and `y` point to the same object, as shown my their identical memory adress
* `x` and `y` are aliases of the same object in memory! 
* Changing `y` therefore changes `x` as well
* [Visual demo on this in Python Tutor](https://pythontutor.com/visualize.html#code=x%20%3D%20%5B1,%202,%203%5D%0Ay%20%3D%20x%0Ay.append%284%29%0Aprint%28x%29%0Aprint%28y%29%0A%0Aprint%28id%28x%29%29%0Aprint%28id%28y%29%29%0Aprint%28id%28x%29%20%3D%3D%20id%28y%29%29%0A&curInstr=8&mode=display&origin=opt-frontend.js&primitivesAsObjects=true&py=311)


## Type hints
* Type hints are a way to help avoid typing errors in Python
* Hints about which data type we expect the function input and output to have
    ```python
    def process_class_scores(scores: list[float], pass_threshold: float = 50.0) -> float:
    ```
* `list[float]` indicates that we expect the input score to be a list of floating-point numbers
* `-> float` indicated the we expect the function to return a floating-point number

````python
def add_no_hints(x, y):
    """Add two numbers together."""
    return x + y


def add_with_hints(x: int | float, y: int | float) -> int | float:
    """Add two numbers together."""
    return x + y


print(add_no_hints(1, 2))  # Output: 3
print(add_no_hints(1.5, 2.5))  # Output: 4.0
print(add_no_hints("hello", " world"))  # Output: "hello world"

print(add_with_hints(1, 2))
print(add_with_hints(1.5, 2.5))
print(add_with_hints("hello", " world"))
````

* Note that the type hints are not enforced!
* We can still input strings to the function that has int and float as type hints
* We should still use type hints when writing Python functions
* They help document the function
* They force you to think through which data types you expect
* They can be used by type checkers like `mypy` to identify potential errors early
* [PEP484](https://peps.python.org/pep-0484/) discuss type hints
* See also chapter 2 of (https://link.springer.com/book/10.1007/978-3-031-40336-1)

### Common type hint patterns
* Single type
    ```python
    def add_ints(x: int, y: int) -> int:
        """Add two integers together."""
        return x + y
    ```
* Multiple types'
    ```python
    def add(x: int | float, y: int | float) -> int | float:
        """Add two numbers together."""
        return x + y
    ```
* Containers
    ```python
    def calculate_grade_stats(
        student_list: list[dict[str, str]], pass_threshold: float = 50.0
    ) -> dict[str, Any]:
    ```
* Special types
    * `Any`
    * `None`
    ```python
    def my_print_function(x: int | float | str) -> None:
        """Print a value."""
        print(x)
    ```


## A deep dive into lists
* A list is a very useful container in Python
* A container can hold and structure a collection of other objects
* Sets, tuples and dicts are some other examples of containers
* A list is a sequence, meaning that the order of elements matter
* It is random access; you can access any element you want in any order
* It is mutable, meaning that is can be changed

### Copying lists

````python
list1 = [1, 2, 3, 4, 5]
list2 = list1

print(f"list1 is at {id(list1)}")
print(f"list2 is at {id(list2)}")
````

* `list1` and `list2` are aliases of the same object
* We can make a shallow copy of a list by using `.copy()` or by slicing `[:]`

````python
list3 = list1.copy()
print(f"list3 is at {id(list3)}")

list4 = list1[:]
print(f"list4 is at {id(list4)}")
````

* These now represent different object in memory
* What happens if we have a nested list?

````python
nested_list1 = [1, 2, [3, 4], 5]
print(f"nested_list1 is at {id(nested_list1)}")

nested_list2 = nested_list1.copy()
print(f"nested_list2 is at {id(nested_list2)}")

nested_list2[2][0] = 0
print(nested_list2)
print(nested_list1)
````

* They are different object, but changing one still changed the other. What happened?
* A shallow copy duplicates the outer container but leaves the inner elements pointing to the same original objects in memory

````python
print(f"nested_list1 element 2 is at {id(nested_list1[2])}")
print(f"nested_list2 element 2 is at {id(nested_list2[2])}")
````
