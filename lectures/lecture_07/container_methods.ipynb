# An Overview of Python Container Methods
Check out Python's documetation for the different containers to learn more about these and other methods.

## List Methods

| Method | Purpose | Returns | Mutates list? |
|:-|:-|:-|:-|
| append(x) | Adds item x to the end of the list | None | Yes |
| extend(iter) | Appends all items from an iterable to the end of the list | None | Yes |
| insert(i,x) | Inserts item x at specified index i | None | Yes |
| remove(x) | Removes first occurrence of value x (raises ValueError if missing) | None | Yes |
| pop(i) | Removes and returns item at index i (default is last item -1) | Item | Yes |
| clear() | Removes all items from the list | None | Yes |
| index(x) | Returns zero-based index of first occurrence of x (raises ValueError if missing) | Integer | No |
| count(x) | Returns total number of times x appears in the list | Integer | No |
| sort() | Sorts list elements in-place (supports optional key and reverse) | None | Yes |
| reverse() | Reverses order of list elements in-place | None | Yes |
| copy() | Returns a shallow copy of the list | New list | No |

## Dict methods

| Method | Purpose | Returns | Mutates dict? |
|:-|:-|:-|:-|
| get(key, default) | Returns value for key if present; otherwise returns default (None if omitted) | Value or default | No |
| keys() | Provides a dynamic view object of all dictionary keys | dict_keys view | No |
| values() | Provides a dynamic view object of all dictionary values | dict_values view | No |
| items() | Provides a dynamic view object of all (key, value) tuples | dict_items view | No |
| update(other) | Updates dict with key-value pairs from another dict or iterable of pairs | None | Yes |
| pop(key, default) | Removes key and returns its value (returns default or raises KeyError if missing) | Value of key | Yes |
| popitem() | Removes and returns the last-inserted (key, value) pair | tuple (key, value) | Yes |
| setdefault(key, default) | Returns value if key exists; if missing, inserts key with default and returns default | Value or default | Yes (if key missing) |
| clear() | Removes all key-value pairs from the dictionary | None | Yes |
| copy() | Returns a shallow copy of the dictionary | dict (new copy) | No |
| fromkeys(iterable, val) | Creates a new dictionary with keys from iterable and values set to val | dict (new dict) | No |

## Set methods

| Method | Purpose | Returns | Mutates set? |
|:-|:-|:-|:-|
| add(elem) | Adds element elem to the set | None | Yes |
| update(others) | Adds all elements from one or more iterables to the set | None | Yes |
| remove(elem) | Removes elem from set (raises KeyError if missing) | None | Yes |
| discard(elem) | Safely removes elem from set if present (no error if missing) | None | Yes |
| pop() | Removes and returns an arbitrary element (raises KeyError if empty) | Removed element | Yes |
| clear() | Removes all elements from the set | None | Yes |
| copy() | Returns a shallow copy of the set | set (new copy) | No |
| union(others) | Returns a new set with elements from current set and all others  | set (new set) | No |
| intersection(others) | Returns a new set with elements common to current set and all others  | set (new set) | No |
| difference(others) | Returns a new set with elements in current set but not in others  | set (new set) | No |
| symmetric_difference(other) | Returns a new set with elements in either set, but not both  | set (new set) | No |
| intersection_update(others) | Updates set in-place, keeping only elements present in all sets  | None  | Yes |
| difference_update(others) | Updates set in-place, removing elements found in specified sets. | None | Yes |
| symmetric_difference_update(other) | Updates set in-place, keeping elements in either set, but not both | None | Yes |
| isdisjoint(other) | Returns True if two sets share zero elements | bool | No |
| issubset(other) | Returns True if every element of current set is in other | bool | No |
| issuperset(other) | Returns True if every element of other is in current set | bool | No |
