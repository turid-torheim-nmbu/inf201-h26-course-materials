## Virtual environments (venv)

* A virtual environment is an well-organised way to manage installation of Python packages
* Each venv is an isolated folder structure where you manage project-specific dependencies <i>independently</i>
* Allows you to keep track of which specific versions of packages are used, which improves reproducibility
* Best practice is to create a new venv for each project you work on
    * You can use the same venv for the first four weeks of exercises
    * Create a new venv for each two-week deliverable project
* VSC will ask you to select (or create) an environment before you run code for the first time in a project
    * Use the command pallette: Ctrl+Shift+P -> Python: Create Environment -> venv
    * Or create one from the terminal: python -m venv .venv
* Use pip to install packages only when the venv is activated
    * This ensures the packages end up in the venv and not your general Python installation
    * Check the terminal to see if your venv has been activated (VSC should do this automatically)
    * Terminal when not activated:
        ``` PS C:\Users\tutor5923\inf201\h26\lectures\lecture_01>```
    * Terminal when activated:
        ``` (.venv) PS C:\Users\tutor5923\inf201\h26\lectures\lecture_01>```

## Write pretty and tidy code: PEP8
* Easy to read, also by someone else than yourself
* Use comments to make you code easier to understand
* Follow the [Python style guideline (PEP8)](https://legacy.python.org/dev/peps/pep-0008/)
* Readability is the main focus of PEP8
* Linters (such as ruff or flake8) help indicate PEP8 violations in your code
* We will look at some examples now, but please read all of PEP8

### Naming conventions
* Functions, variables, methods: named_like_this
* Constants: ALL_CAPS
* Classes: ClassName
* Never use `l`, `O` or `I` for single-letter names (too easy to mistake for `0` or `1`)
* Use informative names

### Spacing
* Not too much whitespace, but also not too little!
* No spaces inside parentheses, brackets, or braces
* Exactly one space around assignment (=) and comparison operators (==, <, >)
    * Not like this:
    ```python 
        total = price * ( 1 + tax_rate )
        items = [ 1, 2, 3 ]
        value=10
    ```
    * But like this:
    ```python
        total = price * (1 + tax_rate)
        items = [1, 2, 3]
        value = 10
    ```

### Indentation and line length
* Indentation in Python is 4 spaces
* Lines should not be longer than 79 characters (in INF201 we allow 120)
* Lines that are too long should be wrapped, using parenthesis to keep things together
* Two blank lines before and after top level functions 

### Imports
* All imports should be at the top of the file, with one line per package
* The order of imports should be as such:
```python
    # 1. Standard library imports
    import os
    import sys
    from typing import Any

    # 2. Third-party imports (external packages)
    import numpy as np
    import pandas as pd

    # 3. Local application imports (your project files)
    from my_project.models import User
    from my_project.utils import format_date
```

### Example of a script that misbehaves

This next piece of code is supposed to calculate the total discount on a shopping cart, given a fixed discount percentage. However, we get the wrong total. Can you identify the error?

````python
def calculate_total_discount(
    prices: list[float], discount_rate: float = 0.10
) -> float:
    """Calculates the total savings across a list of item prices."""
    total_discount = 0.0
    for price in prices:
        item_discount = price * discount_rate
        total_discount += item_discount
        return total_discount


# Sample shopping cart: Total value is $180.00 (expected discount: $18.00)
cart_prices = [100.0, 50.0, 20.0, 10.0]
savings = calculate_total_discount(cart_prices)

print(f"Total savings: ${savings:.2f}")
````
