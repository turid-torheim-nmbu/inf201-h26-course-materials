### Recommended workflow for INF201 deliverables
1. Creating project repository
    * One of you creates the repository in GitHub (webpage), and will be the host for the project
    * Set up the repository like we discussed last week
    * In GitHub, go to Settings - Collaborators - Add people
    * Search for the other collaborator(s) by their GitHub usernames or emails and send invitations
    * They accept the invitation to get write permission to the repository
    * Once everyone has accepted the invitation, everyone can clone the repository to their laptop
    * Note: For small group projects like this, I don't recommend forking. Forking creates disconected repositories that makes everyday collaboration more difficult.
2. Protect `main` and use branches well
    * `main` is (in general) supposed to represent working runnable code
    * It is therefore generally a bad idea to push directly to main with `git push origin main`
    * Use braches for development, and merge them to main when they are working well
    * It is a good idea to have a naming convention for the branches, to make it clear what they are about. Some examples:
        * `fix/file-save-bug` for bug fixes
        * `feature/csv-reader` for developing a new feature in the proeject 
3. Working on a branches and creating pull requests
    * You publish the branch you have created by clicking "Publish Branch" in VS Code or in terminal by typing `git push -u origin <branch>`
    * When you go to GitHub, there will now be a pop-up asking you to "Compare & pull requests"
    * Clicking this will allow you to review the changes in this branch, and tell you if this can successfully be merged with `main`
    * If you want to do more work on the branch before merging with `main`, click the "Create Pull Request" button and select "Create Draft Pull Request"
    * Keep working and commit and push often.
    * When the branch is ready, click the "Ready for review" button at the bottom of the draft pull request
4. Review pull requests
    * If you have assign a collaborator to review the code, they will get a notification about this
    * Note: This should be done by someone who did not write the code!
    * This person then looks over the changes and decide if the merge should be approved, or if changes should be done
    * If the pull request is approved, you can "Merge pull request" and "Confirm merge", which will merge the branch into `main`
    * After the merge, the branch can be deleted to keep the repository tidy
    * See the attached pull request template (generated using Gemini)
5. After merging into main, all collaborators should pull to their own laptop
    * Switch to `main`of you are not already there: `git switch main` (or select main in VS Code)
    * Select "Pull" from the VS Code menus, or on terminal: `git pull`
    * Delete your local version of the branch that has been removed from the origin: `git branch -d <branch-name>`

### Representational State Transfer (REST) 
* [REST](https://en.wikipedia.org/wiki/REST) is a widely used technique for interaction between clients (web browsers, programs using data) and servers
* It is a stateless protocol: the server handles each request independently
* The server offers data through a Web Application Programming Interface (API), i.e., a set of rules describing how to request data and in which form data will be returned
* Data is typically request by a GET request to the server sending a Uniform Resource Identifier (URI), a URL with additional parts to specify which data are requested
    * Example: https://frost.met.no/observations/v0.jsonld?sources=SN17850&elements=mean%28air_temperature+P1D%29&referencetime=2021-08-01%2F2021-09-22'
    * The same with line breaks for clarity:  
        ```
        https://frost.met.no/observations/v0.jsonld
            ?sources=SN1785
                &elements=mean%28air_temperature+P1D%29
                &referencetime=2021-08-01%2F2021-09-22'
    * Server returns data, most often in JSON format, but HTML or XML are also used
    * In Python, we use the package `requests` for communicating with APIs

## Recursive functions
* A [recursive function](https://en.wikipedia.org/wiki/Recursion_(computer_science)) is a function that calls itself
* This breaks the problem down into smaller steps
* A recursive functions needs to parts:
    * Base case: The stopping condition that returns a value without making further recursive calls. Without this the keep just keep running without knowing when to stop
    * Recursive step: The part where the function calls itself with an altered argument moving it closer to the base case.
* Example: Calculating factorials
    * $n! = n*(n-1)*(n-2)* ...*2*1 $
    * We use the fact that $n! = n*(n-1)!$ to create a recursive function

````python
def factorial(n: int) -> int:
    """Calculate the factorial of n"""
    # Base case: stop when n is 1 or less
    if n <= 1:
        return 1

    # Recursive step: call factorial with (n - 1)
    return n * factorial(n - 1)


print(factorial(5))  # Output: 120
````

## Character encoding
* How does the computer view a string or the contents of a file?

### [ASCII](https://en.wikipedia.org/wiki/ASCII)
* American Standard Code for Information Interchange
* From the 1960s
* 7 bit, numerical values 0-127

    |Range|Hex|Values|
    |--:|--:|:--|
    |0-31|0x00-0x1f|Control characters, e.g., \"newline\"|
    |32|0x20|Space|
    |33-47|0x21-02f|Punctuation|
    |48-57|0x30-0x39|0-9|
    |58-64|0x3a-0x40|Punctuation|
    |65-90|0x41-0x5a|A-Z|
    |91-96|0x5b-0x60|Punctuation|
    |97-122|0x61-0x7a|a-z|
    |123-126|0x7b-0x7e|Punctuation|
    |127|0x7f|Control character DEL|

### [Unicode](https://en.wikipedia.org/wiki/Unicode) UTF-8
* ISO 10646 standard
* Provides 1,114,112 code points which can represent one symbol each
* Currently, 159,801 code points defined
    * 154 scripts/languages
    * mathematics
    * emojis, ...
* Code points up to 255 (0xff) agree with ASCII (to 127) and [ISO 8859-1 Latin 1](https://en.wikipedia.org/wiki/ISO/IEC_8859-1) for 128-255
* Code points are abstract numbers
* Code points are written as U+005a (capital "Z")
* Actual storage: Unicode Transformation Format (UTF)
    * Most common: UTF-8 (almost 99% of all websites)
    * UTF-8 is default for Python
    * Also used: UTF-16 and UTF-32
    * UTF-8 encodes the number of a Unicode code point in 1, 2, 3 or 4 bytes
    * UTF-8 is identical to ASCII up to 127
    * Number of bytes and bits for different code point (CP) ranges (after [Wikipedia](https://en.wikipedia.org/wiki/UTF-8)):

        |Bytes|Bits for CP|First CP|Last CP|Byte 1|Byte 2|Byte 3|Byte 4|
        |-:|-:|-:|-:|-:|-:|-:|-:|
        |1|7|U+0000|U+007F|`0xxxxxxx`||||
        |2|11|U+0080|U+07FF|`110xxxxx`|`10xxxxxx`|||
        |3|16|U+0800|U+FFFF|`1110xxxx`|`10xxxxxx`|`10xxxxxx`||
        |4|21|U+10000|U+10FFFF|`11110xxx`|`10xxxxxx`|`10xxxxxx`|`10xxxxxx`|

### Character encoding in Python

### How do I know which encoding a file has?
* This is more difficult than it sounds!
* Plain text files do not always contain built-in metadata stating their encoding.
* They will sometimes have a [Byte Order Mark (BOM)](https://en.wikipedia.org/wiki/Byte_order_mark)
    * A specific sequence of bytes placed at the very beginning of a file to signal its Unicode encoding and byte order (endianness)
    * Some of the most common BOMs:

        |Encoding|BOM (hex)|Python Byte String|
        |-:|-:|-:|
        |UTF-8|`EF BB BF`|`b"\xef\xbb\xbf"`|
        |UTF-16 (little-endian)|`FF FE`|`b"\xff\xfe"`|
        |UTF-16 (big-endian)|`FE FF`|`b"\xfe\xff"`|

* When trying to figure out what encoding a file has, you can read the file in byte mode and look for a BOM

````python
from pathlib import Path

file_path = Path("file_utf8_with_bom.txt")
raw_bytes = file_path.read_bytes()
raw_bytes
````

* This file has the UTF-8 BOM
* Now, let's look at a similar file without a BOM

````python
file_path = Path("file_utf8_no_bom.txt")
raw_bytes = file_path.read_bytes()
raw_bytes
````

* Suggested method for identifying the encoding of a file:
    1. Check for BOM: If present, read the file with encoding `UTF-8-sig` or `UTF-16` (or `UTF-32`)
    2. Attempt to read the file with UTF-8 encoding
    3. IF this fails (`UnicodeDecodeError`), try reading it with `cp1252` or Latin-1 encoding
        * `cp1252`is almost identical to Latin-1, and favoured by Windows
* The Python package [Charset normalizer](https://pypi.org/project/charset-normalizer/) will analyse the file and predict the character encoding
* Even if Unicode in the form of UTF-8 is the most common system, there are a lot of different character encoding systems
    * [This is an overview of all encodings supported in Python](https://docs.python.org/3/library/codecs.html#standard-encodings)

````python
from charset_normalizer import from_path

results = from_path("file_utf8_no_bom.txt")
best_match = results.best()
print("Detected Encoding:", best_match.encoding)
````

* Why did this say ascii and not UTF-8?
* Most likely because the file is very small, and therefore don't have enough content to make a good guess
* What if we instead check the file containing a BOM?

````python
results = from_path("file_utf8_with_bom.txt")
best_match = results.best()
print("Detected Encoding:", best_match.encoding)
````

### Encoding and decoding
* Encoding: converting a str to its bytes representation
* Decoding: Converting from some encoding to Python `str`
* Strange things can happen if we choose the wrong encoding to decode with
* UTF-8 is default encoding

````python
text = "NMBU er i Ås kommune."
text_utf8 = bytes(text, encoding='utf-8')
text_latin1 = bytes(text, encoding='latin1')
print('UTF-8  :', text_utf8)
print('Latin-1:', text_latin1)
````

* Note the different encodings: UTF-8 requires two bytes for each Norwegian character
* Now, let's try to decode this again

````python
text_utf8.decode()
````

* As UTF-8 is the default, this worked well
* What about the Latin-1 encoded string?

````python
text_latin1.decode()
````

* We can tell `decode()` how to handle errors

````python
text_latin1.decode(errors='replace')
````

````python
text_latin1.decode(errors='ignore')
````

````python
text_latin1.decode(encoding='latin1')
````

````python
text_utf8.decode(encoding='latin1')
````
